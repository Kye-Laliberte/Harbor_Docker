```python
from app.services.travel_time_service import ShipSpeedRegression

# Completed voyages are training labels:
# (ship_id, route_distance_km, ship_size, observed_speed_kmh)
training_rows = [
    (1, 800.0, 2, 22.0),
    (1, 1200.0, 2, 21.0),
    (1, 1800.0, 2, 20.0),
    (2, 800.0, 2, 31.0),
    (2, 1200.0, 2, 30.0),
    (2, 1800.0, 2, 29.0),
    (3, 1000.0, 3, 18.0),
    (3, 1600.0, 3, 17.5),
]

model = ShipSpeedRegression(
    learning_rate=0.01,
    epochs=5000,
    regularization=1e-3,
    tolerance=1e-9,
).fit(training_rows)

route_distance_km = 1500.0
ship_predictions = []
for ship_id, ship_size in [(1, 2), (2, 2), (3, 3), (99, 2)]:
    predicted_speed_kmh = model.predict(route_distance_km, ship_size, ship_id)
    ship_predictions.append(
        {
            "ship_id": ship_id,
            "predicted_speed_kmh": round(predicted_speed_kmh, 2),
            "average_voyage_time_hours": round(route_distance_km / predicted_speed_kmh, 2),
            "ship_training_samples": model.ship_sample_count(ship_id),
        }
    )

print(f"Model fitted: {model.is_fitted}")
print(f"Optimizer: batch_gradient_descent")
print(f"Epochs: {model.epochs_run}")
print(f"Training loss: {model.training_loss:.6f}")
print(f"Route distance: {route_distance_km:.0f} km")
print()
for prediction in ship_predictions:
    print(prediction)
```
